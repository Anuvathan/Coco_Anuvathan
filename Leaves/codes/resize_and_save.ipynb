```python
import os
import cv2
import shutil
```

```python
input_dir = '../dataset'
output_dir = '../dataset_resized_512x512/'
```

```python
# Create output directories if they don't exist
os.makedirs(os.path.join(output_dir, 'healthy'), exist_ok=True)
os.makedirs(os.path.join(output_dir, 'leaf_spot'), exist_ok=True)
os.makedirs(os.path.join(output_dir, 'caterpillar'), exist_ok=True)
os.makedirs(os.path.join(output_dir, 'leaf_blight'), exist_ok=True)
os.makedirs(os.path.join(output_dir, 'rhinoceros'), exist_ok=True)
```

```python
# Loop through each category and process the images
categories = ['healthy', 'leaf_spot', 'caterpillar', 'leaf_blight', 'rhinoceros']

for category in categories:
    category_input_dir = os.path.join(input_dir, category)
    category_output_dir = os.path.join(output_dir, category)
    
    # List all files in the input directory
    image_files = os.listdir(category_input_dir)
    
    for image_file in image_files:
        input_path = os.path.join(category_input_dir, image_file)
        output_path = os.path.join(category_output_dir, image_file)
        
        # Load the image using OpenCV
        image = cv2.imread(input_path)
        
        if image is not None:
            # Resize the image to your desired dimensions (e.g., 224x224)
            resized_image = cv2.resize(image, (512, 512))
            
            # Save the resized image to the output directory
            cv2.imwrite(output_path, resized_image)
```
