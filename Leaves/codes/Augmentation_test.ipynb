```python
import tensorflow as tf
import matplotlib.pyplot as plt

# Load the original image
original_image_path =  '../dataset_resized_512x512/healthy/IMG_0024.JPG' # Replace with the path to your image
original_image = tf.io.read_file(original_image_path)
original_image = tf.image.decode_image(original_image, channels=3)

# Horizontally flip the original image
flipped_image = tf.image.flip_left_right(original_image)

# Rotate the original and flipped images by 90 and -90 degrees
rotated_image_90 = tf.image.rot90(original_image)
rotated_image_neg90 = tf.image.rot90(flipped_image)

# Define brightness and contrast adjustment factors
brightness_factors = [-0.2, -0.1, 0.1, 0.2]
contrast_factors = [0.8, 0.9, 1.1, 1.2]

# Apply brightness and contrast adjustments to all images
augmented_images = []

for brightness_factor in brightness_factors:
    for contrast_factor in contrast_factors:
        augmented_image = tf.image.adjust_brightness(original_image, brightness_factor)
        augmented_image = tf.image.adjust_contrast(augmented_image, contrast_factor)
        augmented_images.append(augmented_image)

# Plot the original and augmented images
plt.figure(figsize=(15, 15))

images_to_plot = [original_image, flipped_image, rotated_image_90, rotated_image_neg90] + augmented_images
titles = ['Original', 'Flipped', 'Rotated 90°', 'Rotated -90°'] + [f'Brightness {b}, Contrast {c}' for b in brightness_factors for c in contrast_factors]

for i, image in enumerate(images_to_plot):
    plt.subplot(4, 7, i + 1)
    plt.imshow(image.numpy().astype(int))
    plt.title(titles[i])
    plt.axis('off')

plt.show()
```
