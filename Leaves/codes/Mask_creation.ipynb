```python
import cv2
import numpy as np
import os

# Specify the directory where your images are located
input_directory = '../dataset/segmented/caterpillar/'

# Specify the directory where you want to save the mask images
output_directory = '../dataset/segmented/caterpillar/mask/'

# Make sure the output directory exists, create it if necessary
if not os.path.exists(output_directory):
    os.makedirs(output_directory)

# List all the PNG files in the input directory
input_files = [f for f in os.listdir(input_directory) if f.endswith('.png')]

# Process each image
for input_file in input_files:
    # Load the image with transparency
    image = cv2.imread(os.path.join(input_directory, input_file), cv2.IMREAD_UNCHANGED)
    
    # Create the mask
    mask = np.where(image[:, :, 3] == 0, 0, 255)
    
    # Get the original image's filename without the extension
    filename_without_extension = os.path.splitext(input_file)[0]
    
    # Construct the output filename for the mask image
    output_mask_filename = os.path.join(output_directory, filename_without_extension + '_mask.png')
    
    # Save the mask image
    cv2.imwrite(output_mask_filename, mask)

print("Mask images saved successfully.")
```
