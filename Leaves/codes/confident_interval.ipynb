```python
import numpy as np
import scipy.stats as stats

# Replace this with your list of accuracy scores
accuracy_scores = [0.531278,	0.589343,	0.549534,	0.579002,	0.543468]

# Calculate the mean accuracy
mean_accuracy = np.mean(accuracy_scores)

# Calculate the standard deviation
std_deviation = np.std(accuracy_scores)

# Number of folds (cross-validation splits)
n = len(accuracy_scores)

# Calculate the margin of error (MOE) for a 95% confidence interval
z_score = stats.norm.ppf(0.975)  # 1.96 for a 95% confidence interval
MOE = (std_deviation / np.sqrt(n)) * z_score

# Calculate the lower and upper bounds of the confidence interval
lower_bound = mean_accuracy - MOE
upper_bound = mean_accuracy + MOE

print(f"Mean Accuracy: {mean_accuracy:.6f}")
print(f"STD Accuracy: {std_deviation:.6f}")
print(f"Final Accuracy: {mean_accuracy * 100:.4f}% ± {MOE * 100:.4f}%")
print(f"Confidence Interval: ({lower_bound * 100:.4f}%, {upper_bound * 100:.4f}%)")
```
