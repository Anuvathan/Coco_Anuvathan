```python
import numpy as np
from sklearn.model_selection import KFold, train_test_split

# Set the random state for reproducibility
random_state = 13

# Create an array of image IDs from 1 to 165
image_ids = np.arange(1, 119)

# Initialize KFold with 5 folds and the specified random state
kf = KFold(n_splits=5, random_state=random_state, shuffle=True)

# Initialize a list to store the image IDs in each fold
fold_image_ids = []

# Perform the random split into 5 folds
for train_index, test_index in kf.split(image_ids):
    train_fold = image_ids[train_index]
    test_fold = image_ids[test_index]

    # Split the train set into train and validation sets (80% train, 20% validation)
    train_ids, val_ids = train_test_split(train_fold, test_size=0.2, random_state=random_state)

    fold_image_ids.append({
        "train": np.sort(train_ids).tolist(),
        "val": np.sort(val_ids).tolist(),
        "test": np.sort(test_fold).tolist()
    })

# Print the image IDs in each fold
for fold_num, fold_data in enumerate(fold_image_ids):
    print(f"Fold {fold_num + 1}:")
    print(f"Train Set: {fold_data['train']}")
    print(f"Validation Set: {fold_data['val']}")
    print(f"Test Set: {fold_data['test']}")
    print()
```
