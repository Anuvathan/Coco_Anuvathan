```python
import cv2
import os
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Load the image
image_path = 'IMG_0663.jpg'
image = cv2.imread(image_path)
height, width, _ = image.shape
```

```python
# Extract the filename without extension
filename_without_extension = os.path.splitext(os.path.basename(image_path))[0]

# Reshape the image into a 2D array of pixels
pixel_values = image.reshape((-1, 3))

# Specify the number of clusters (in your case, 3)
num_clusters = 3

# Perform K-means clustering
kmeans = KMeans(n_clusters=num_clusters)
kmeans.fit(pixel_values)

# Get the cluster labels
cluster_labels = kmeans.labels_

# Create segmented images
segmented_images = []

for i in range(num_clusters):
    mask = (cluster_labels == i).reshape(height, width)
    segmented_image = np.zeros_like(image)
    segmented_image[mask] = image[mask]
    segmented_images.append(segmented_image)

# Create a subplot with 2 rows and 2 columns
fig, axs = plt.subplots(2, 2, figsize=(10, 10))

# Plot the original image
axs[0, 0].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
axs[0, 0].set_title('Original Image')

# Plot the segmented images
for i, seg_image in enumerate(segmented_images):
    axs[(i + 1) // 2, (i + 1) % 2].imshow(cv2.cvtColor(seg_image, cv2.COLOR_BGR2RGB))
    axs[(i + 1) // 2, (i + 1) % 2].set_title(f'Segmented Image {i+1}')

# Hide axis labels and ticks
for ax in axs.flat:
    ax.axis('off')

# Display the plot
plt.tight_layout()
plt.show()
```

```python
# Save the segmented images
for i, seg_image in enumerate(segmented_images):
    seg_filename = f'{filename_without_extension}_seg{i+1}.jpg'
    cv2.imwrite(seg_filename, seg_image)

print("Segmented images saved")
```
